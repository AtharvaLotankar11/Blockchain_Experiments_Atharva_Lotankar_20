```python
import datetime
import hashlib
import json

class Blockchain:

    def __init__(self):
        self.chain = []
        self.transactions = []
        # Create Genesis Block (Block 1)
        self.create_block(proof=1, previous_hash='0')

    def create_block(self, proof, previous_hash):
        """
        Creates a new block in the blockchain and attaches current transactions.
        Clears pending transactions list afterward.
        """
        block = {
            'index': len(self.chain) + 1,
            'timestamp': str(datetime.datetime.now()),
            'proof': proof,
            'previous_hash': previous_hash,
            'transactions': self.transactions
        }
        self.chain.append(block)
        # Empty transaction list after mining
        self.transactions = []
        return block

    def get_previous_block(self):
        return self.chain[-1]

    def create_transaction(self, sender, receiver, amount):
        """
        Adds transaction to pending list and returns target block index.
        """
        self.transactions.append({
            'sender': sender,
            'receiver': receiver,
            'amount': amount
        })
        return len(self.chain) + 1

    def proof_of_work(self, previous_proof):
        """
        Cryptographic puzzle: searches for a golden nonce where
        SHA-256 output starts with '000' leading zeros.
        """
        new_proof = 1
        check_proof = False

        while check_proof is False:
            hash_operation = hashlib.sha256(
                str(new_proof**2 - previous_proof**2).encode()
            ).hexdigest()

            if hash_operation[:3] == '000':
                check_proof = True
            else:
                new_proof += 1

        return new_proof

    def hash(self, block):
        encoded_block = json.dumps(block, sort_keys=True).encode()
        return hashlib.sha256(encoded_block).hexdigest()

    def is_chain_valid(self, chain):
        previous_block = chain[0]
        block_index = 1

        while block_index < len(chain):
            block = chain[block_index]

            # 1. Verify previous hash matching
            if block['previous_hash'] != self.hash(previous_block):
                return False

            # 2. Verify Proof of Work target (3 leading zeros)
            previous_proof = previous_block['proof']
            proof = block['proof']
            hash_operation = hashlib.sha256(
                str(proof**2 - previous_proof**2).encode()
            ).hexdigest()

            if hash_operation[:3] != '000':
                return False

            previous_block = block
            block_index += 1

        return True

print("Blockchain class successfully defined!")
```

```python
# Instantiate Blockchain
blockchain = Blockchain()

def mine_block():
    # Mine ONLY if transaction list is not empty
    if not blockchain.transactions:
        return {'message': 'No transactions to mine. Add some transactions first!'}

    previous_block = blockchain.get_previous_block()
    previous_proof = previous_block['proof']
    proof = blockchain.proof_of_work(previous_proof)
    previous_hash = blockchain.hash(previous_block)

    block = blockchain.create_block(proof, previous_hash)

    return {
        'message': 'Congratulations, you just mined a block!',
        'index': block['index'],
        'timestamp': block['timestamp'],
        'proof': block['proof'],
        'previous_hash': block['previous_hash'],
        'transactions': block['transactions']
    }

def add_transaction(sender=None, receiver=None, amount=None):
    if sender is None:
        sender = input("Enter sender: ")
        receiver = input("Enter receiver: ")
        amount = float(input("Enter amount: "))

    index = blockchain.create_transaction(sender, receiver, amount)
    return {'message': f'Transaction will be added to Block {index}'}

def get_chain():
    return {
        'chain': blockchain.chain,
        'length': len(blockchain.chain)
    }

def is_valid():
    valid = blockchain.is_chain_valid(blockchain.chain)
    if valid:
        return {'message': 'All good. The Blockchain is valid.'}
    return {'message': 'Houston, we have a problem. The Blockchain is not valid.'}

print("Core functions initialized successfully!")
```

```python
# Block 1 is already auto-generated as Genesis Block

# Add Transaction & Mine Block 2
print(add_transaction(sender="Atharva", receiver="Manu", amount=100.0))
print(json.dumps(mine_block(), indent=2))
print("-" * 50)

# Add Transaction & Mine Block 3
print(add_transaction(sender="Manu", receiver="Ronit", amount=45.5))
print(json.dumps(mine_block(), indent=2))
print("-" * 50)

# Add Transaction & Mine Block 4
print(add_transaction(sender="Ronit", receiver="Me", amount=12.0))
print(json.dumps(mine_block(), indent=2))
```
